```python
def upper_confidence_bound(mu, sigma, kappa=2.0):
    """
    Upper Confidence Bound (UCB) acquisition function.
    UCB = mean + kappa * std
    Parameters:
    -----------
    mu : predicted mean
    sigma : predicted standard deviation
    kappa : exploration parameter (higher = more exploration)
    """
    return mu + kappa * sigma

def lower_confidence_bound(mu, sigma, kappa=2.0):
    """
    Lower Confidence Bound (UCB) acquisition function.
    LCB = mean - kappa * std
    Parameters:
    -----------
    mu : predicted mean
    sigma : predicted standard deviation
    kappa : exploration parameter (higher = more exploration)
    """
    return mu - kappa * sigma

def expected_improvement(mu, sigma, y_best, xi=0.01):
    """
    Expected Improvement (EI) acquisition function.
    EI = E[max(f(x) - f(x_best), 0)]
    Parameters:
    -----------
    mu : predicted mean
    sigma : predicted standard deviation  
    y_best : best observed value so far
    xi : exploration parameter
    """
    with np.errstate(divide='warn'):
        improvement = mu - y_best - xi
        Z = improvement / sigma
        ei = improvement * norm.cdf(Z) + sigma * norm.pdf(Z)
        ei[sigma == 0.0] = 0.0
    return ei
```

```python
def acq_objective(X):
    X = X.reshape(1, -1)
    mu, sigma = gp.predict(X, return_std=True)
               
    if acquisition == 'ei':
        acq_val = expected_improvement(mu, sigma, Y.max())
    elif acquisition == 'lcb':
        acq_val = lower_confidence_bound(mu, sigma, kappa=1.25)
    else:
        acq_val = upper_confidence_bound(mu, sigma, kappa=2.5)
           
    return -acq_val  # Negative because we use the 'minimize' function later
```

```python
acquisition='lcb'
n_dims=X.shape[1]
bounds = [(0, 1) for _ in range(n_dims)]
best_acq = np.inf
X_next = None

for _ in range(300):  # Multiple random starts
    x0 = np.random.uniform([b[0] for b in bounds], [b[1] for b in bounds])
    result = minimize(acq_objective, x0, bounds=bounds, method='L-BFGS-B')
    if result.fun < best_acq:
        best_acq = result.fun
        X_next = result.x

#np.set_printoptions(suppress=True, precision=6)
print(f'New query: {X_next[0]:.6f}-{X_next[1]:.6f}')
# New query: 0.445257-0.336729 - lcb
# New query: 0.370965-0.537526 - lcb
# New query: 0.454384-0.363326 - lcb - k=2.5 - 12/08/26
# New query: 0.809754-0.060363 - ei - xi = 0.01 - 12/08/26
# New query: 0.658648-0.726813 - ei - xi = 0.1 - Anisotropic Mattern2.5 - 13.08.26
# New query: 0.704059-0.493643 - ei - xi = 0.1 - Isotropic Mattern2.5 - 13.08.26
# New query: 0.435783-0.348446 - lcb - k=2.5 - Anisotropic Mattern2.5 - 13.08.26
# New query: 0.548003-0.519056 - lcb - k=2.5 - Isotropic Mattern2.5 - 13.08.26 
# New query: 0.168691-0.769922 - ei - xi = 0.1 - Isotropic Mattern2.5 - 19.08.26 
# New query: 0.316252-0.259798 - lcb - k=1.0 - Isotropic Mattern2.5 - 19.08.26 

# New query: 0.150125-0.191440 - ei - xi = 0.1 - Isotropic Mattern2.5 - 27.08.26  
# New query: 0.364973-0.435463 - lcb - k=1.25 - Isotropic Mattern2.5 - 27.08.26 
```
